```python
def skip_gram(words):
    res = []
    for indx in range(len(words)):
        if(indx == 0):
            res.append([words[0],words[1]])
        elif(indx == len(words)-1):
            res.append([words[-1],words[-2]])
        else:
            res.append([words[indx],words[indx+1]])
            res.append([words[indx],words[indx-1]])
    print(res)

def skip_gram_window(words, window):
    res = []

    for indx in range(len(words)):

        if indx == 0:
            for i in range(indx + 1, window + 1):
                res.append([words[indx], words[indx + i]])

        elif indx <= window:
            for i in range(0, indx):
                res.append([words[indx], words[i]])

            for i in range(indx + 1, indx + window + 1):
                if i < len(words):
                    res.append([words[indx], words[i]])

        elif len(words) - indx <= window:
            for i in range(indx - window, indx):
                res.append([words[indx], words[i]])

            for i in range(indx + 1, len(words)):
                res.append([words[indx], words[i]])

        else:
            for i in range(indx - window, indx):
                res.append([words[indx], words[i]])

            for i in range(indx + 1, indx + window + 1):
                res.append([words[indx], words[i]])

    return res
```
